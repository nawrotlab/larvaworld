## About this notebook

This is the only lesson in the course with no code in it. It states the position Larvaworld takes
on what it means to model a behaving animal, and it is worth ten minutes before you start
configuring simulations - because that position is what the software's structure follows.

The claim is short : a behavioral model is not complete unless it addresses five things -
**energetics**, the **homeostatic interface**, the **nervous system**, the **sensorimotor
interface**, and the **environment**. Each can be modeled at whatever level of abstraction your
question needs, but none can simply be left out without the model quietly becoming an
input-output mapping rather than an organism.

Every one of the five has a home in the package, and the last section of this notebook says where.

**Prerequisites** : none.

**Cost** : reading time, about 15 minutes.

## Section 1 : Introduction

A living organism is an autopoietic system that persists in time, embodied in a physical body and
embedded in its habitat. It constantly evades thermodynamic equilibrium by actively maintaining a
network of biochemical processes (metabolism) that reassure its (re)generation and reproduction. It
accomplishes this by interacting with its environment, exchanging energy/matter, so that it keeps a
set of internal parameters - e.g. glucose, O2, CO2 levels, fluid pressure and osmolality - within
viable boundaries (homeostasis). In this sense, it can be conceptualized as a control system, while
its adaptivity determines its degrees of freedom.

Discrete subsystems address its self-regulation and its agential engagement in the face of
environmental risks and challenges. Regarding self-regulation, two main subsystems coordinate
processes over the whole body, namely the immune and the endocrine/peptide systems. The former
preserves a dynamic boundary between what is part of the self and what is not, while the latter
guides adaptation of internal processes to external change, accomplishing allostasis. On the other
hand, motile parts, from cilia to muscles, allow e.g. ingestion of food or reproductive
performance. Evolution of the nervous system extended both - more specifically mainly the second
but also the latter part of the first - allowing faster, more efficient and proactive processes
giving rise to behavior. Therefore, the nervous system can be regarded as mediating, by extending
them, part of the homeostatic/allostatic control processes of the living metabolism.

This approach leads to two operationalizations that are extremely useful for modeling purposes.

**First**, clusters of processes that have a common homeostatically relevant goal can be regarded
as semi-independent subsystems (*survival circuits*). Such a system integrates both
self-regulation and neurally mediated behavior, coordinating dynamic control towards
integrity/survival, maintenance/development or reproduction.

**Second**, such a system can be dissected into two discrete networks of processes, namely a
metabolically-mediated core and a neurally-mediated behavioral extension. These networks interact
among themselves and with the environment, giving rise to the **homeostatic** interface (between
metabolism and behavior) and the **sensorimotor** interface (between behavior and environment).
Modulation of behavior by metabolic state - homeostatic drive - across the homeostatic interface
allows alignment of behavior towards the accomplishment of homeostatic control. Equivalently, body
coupling to the environment through nested recursive sensorimotor loops across the sensorimotor
interface enables behavioral interactions.

Thus, under the whole-organism framework, behavioral modeling, at whatever level of abstraction,
cannot be considered complete - not in terms of realistic depth, but in terms of normative width -
if it does not address at least these five elements : metabolism/energetics, homeostatic interface,
nervous system, sensorimotor interface and environment. To this end, modeling decisions have to be
made regarding each of them, paying special attention to the interfaces : on the one hand, what
aspects of behavior are homeostatically modulated and how; on the other, how behavior is realized
by the body. The main components - energetics, nervous system and environment - can then be modeled
at a suitable level of abstraction. This approach yields closed-loop normative models of the
behavior serving a defined aspect of autopoiesis, enabling a dynamic mechanistic explanation both
in terms of *how* and *what for*.

## Section 2 : The five essential components

A brief description of each is given below. Each can be modeled at a different level of abstraction
depending on the modeling goals. Moreover, different temporal timescales can be integrated : a
homeostatically relevant timescale of hours or days, a behaviorally relevant timescale of seconds,
and a neurally relevant timescale of milliseconds.

### 1. Energetics

For the purpose of modeling behavior serving energy/matter exchange or reproduction, a suitable
abstraction for the relevant core metabolism network is at the level of energetic flux and
investment. Dynamic energy budget (DEB) theory provides a coherent, biologically plausible starting
point.

According to DEB, the life history of an organism - including transitions across subsequent life
stages - can be captured through the dynamics of four state variables (structural volume, energy
reserve, maturity investment and reproduction buffer), while its interaction with the world and its
self-regulation is described through the dynamics of energy/mass fluxes (assimilation,
mobilization, maturation and maintenance).

Realistic, experimentally testable dynamics of life-history traits such as growth curves can be
simulated via species-specific DEB implementations, whose parameters are approximated using
experimental observations.

### 2. Homeostatic interface

The nervous system senses aspects of homeostatic state - or, equivalently, metabolism modulates
aspects of the nervous system - through interoception and neuromodulation. Be it at the level of
specific chemical compounds or at the abstract level of the balance between global multi-compound
chemical signatures, the nervous system is driven towards or away from certain behavioral patterns.
Three questions have to be answered in this context :

1. What aspect of metabolism/energetics is sensed, and with what sensitivity and dynamics?
2. What aspect of the nervous system is modulated, and with what gain and dynamics?
3. What degrees of freedom does this modulation have, given that its bodily realization raises
   physical constraints?

Assumptions answering these questions guide modeling decisions and define what we call a
**homeostatic drive function**.

### 3. Nervous system

The nervous system can be operationalized as a layered architecture in which every higher layer can
only minimally alter the dynamics of the underlying ones - that is, with only a few degrees of
freedom available relative to the total degrees of freedom of the underlying layers. This approach
stems from the fundamentally decentralized nature of the nervous system, and from the abundant
evidence that simpler behaviors can be carried out by local neural mechanisms even in the absence
(lesion or inhibition) of higher-order neural components. Therefore, reflexive sensorimotor
couplings at the sensorimotor interface are combined into more extended stereotyped reactive loops;
these in turn allow adaptive inhibition/disinhibition of motor primitives by higher layers.
Finally, learning and memory allow persistent long-term behavioral adaptations.

The layered architecture of behavioral processes, alongside their underlying neural mechanisms,
facilitates behavioral modeling because it provides a concrete rule for model evaluation, as
instructed by behavior-based robotics : lower behavioral layers are calibrated and evaluated first,
and only after their basic performance is established can modulation by higher layers be
implemented.

### 4. Sensorimotor interface

The neurally-controlled body interacts constantly with the environment. The level at which this
interaction is abstracted determines three crucial aspects of the overall model :

1. The level of abstraction of the environment.
2. The bottom layer of the layered architecture of the nervous system.
3. The body implementation, ranging from abstract and non-spatial to multi-segment and
   neuromechanically detailed.

In other words, the sensorimotor interface determines the basic behavioral patterns that, in the
context of the specific model, are considered primitives and cannot be dissected further.

### 5. Environment

The environment provides the missing arc that completes the closed-loop dynamic model of a
behavior, which might otherwise be modeled as a mere input-output half-loop. It must therefore
incorporate the dynamic environmental parameters that are relevant to the behavior. Central
questions are :

1. Is space modeled? At what spatial resolution?
2. What is the behavioral temporal resolution?
3. What modalities are modeled - olfaction, vision, touch, audition? At what level of abstraction?
   With what sensitivity and dynamics?
4. Are other agents modeled, e.g. as indirect competition for resources?
5. Is inter-agent (social) interaction allowed, e.g. communication or direct competition?
